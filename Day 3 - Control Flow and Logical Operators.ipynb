```python
#Comparison Operators

print(3 > 2)
print(3 >= 3)
print(1 != 2)
print(3 < 2)

#Comparing data of different types
print(3 == "3")
```

```python
#if / else statement

height = float(input("What is your height in m? "))

if height >= 1.5:
    print("You meet the height requirement for the ride!")
else:
    print("Sorry, you do not meet the height requirement.")
```

```python
#elif statement

age = int(input("How old are you? "))

if age < 12:
    bill = 5
elif age <= 18:
    bill = 7
else:
    bill = 12

print(f"Your ticket costs ${bill}.")
```

```python
#Logical Operators - and, or, not

is_sunny = True
temperature = 28

if is_sunny and temperature > 25:
    print("Great day for the beach!")

is_raining = False
has_umbrella = True

if is_raining or not has_umbrella:
    print("Take an umbrella just in case.")
else:
    print("No umbrella needed.")
```

```python
#Nested if statements

weight = float(input("Weight (kg): "))
height = float(input("Height (m): "))

bmi = weight / height ** 2

if bmi < 18.5:
    print(f"Your BMI is {bmi:.1f}, you are underweight.")
elif bmi < 25:
    print(f"Your BMI is {bmi:.1f}, you have a healthy weight.")
else:
    if bmi < 30:
        print(f"Your BMI is {bmi:.1f}, you are overweight.")
    else:
        print(f"Your BMI is {bmi:.1f}, you are obese.")
```
